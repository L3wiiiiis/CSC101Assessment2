## Problem 1: Tip Calculator
1.1.1 The Scenario
You work at a Perth restaurant. The manager wants a simple tool where staff can enter:

- The bill amount (e.g., $45.80)
- The tip percentage (e.g., 15%)

And get back:

- The tip amount
- The total amount to charge


1.1.2 What You Must Submit

STEP 1: Understand the Problem

In your own words, what is this program supposed to do? (2-3 sentences)

[ The program helps staff in Perth restaurant to work out the tip on customer's bill by typing the bill amount and tip percentage. The program will then show how much tip is supposed to be given and full amount to charge the customer (Bill plus tip) ]



STEP 2: Identify Inputs and Outputs

INPUTS:

[What does the user provide?]
- The bill amount; the cost of the order before the tip
- Tip percentage; used to calculate the actual tip charge

[What type of data? Numbers? Text?]
- The bill amount is a number 
- Tip percentage is a number 

OUTPUTS:

[What does the program show back?]
- Tip amount; the actual tip charge
- Total charge; the total amount to charge the customer
[What format?]
- Both outputs are in form of money rounded to 2 decimal places. 




STEP 3: Work It By Hand

Show me you can do this with a calculator. Use this example:

Bill amount: $45.80
Tip percentage: 15%
Write the steps:



[First calculation I would do]
- Change the 15% into a decimal. 15% is same as 15/100 which is equal to 0.15

[Second calculation]
- Multiply the bill amount ($45.80) by the tip percentage (0.15) to get the actual tip charge.

 $45.80 * 0.15 = $6.87


[Final step]

Add the tip ($6.87) to the bill ($45.80) to get the total amount to charge the customer.

$45.80 + $6.87 = $52.67

[What I would tell the staff member]
- The tip is $6.87 and the total charge is $52.67

OUTPUTS:
