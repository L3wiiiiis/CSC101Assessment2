```python
filename = "my_data.txt"
#creates a variable called filename.
try:
    #shows an attempt of code and if something goes wrong it does not crash.

    with open(filename, "r") as file:
#"with" automatically closes the file when you are done.
#"open(filename, "r")" opens the file specified in filename but in read mode. "r" means read mode.

        content = file.read()
#reads the entire contents of the file.

    lines = content.splitlines()
#puts the text in list.

    print("=== MY FAVOURITE MOVIES ===")
#prints the title.
    print(content)
#prints out the whole file.

#Number of words in the file
    word_count = sum(len(line.split()) for line in lines)
#"sum" adds all the counts together
#"len" counts the words in the list
#"line.split()" breaks one line into words where there is space
#"for line in lines" does the above code for every line

    print("Number of words:", word_count)
# prints out the number of words 

    if lines:
#checks the list has somrthing in it.
        longest_line = max(lines, key=len)
# "max" picks the largest item.
# "key=len" compares number of characters.
        print("\nLongest line:")
#"\n" adds a blank line first so that the output is easier to read.
        print(longest_line)
#shows the line itself
        print("Length of longest line:", len(longest_line))
#shows the character count
    else:
        print("\nThe file is empty.")
# "/n" means new line the file is empty will be printed out

except FileNotFoundError:
# if python could not find the file with that name
    print(f"File '{filename}' not found.")
# prints out the filename (my_data.txt) not found
except Exception as e:
# "Exception" is a general category that covers almost every kind of error
#"as e", the e is a variable for the error message
    print(f"An error occurred: {e}")
# the error meessage is printed out if that except Exception block actually runs.
```
