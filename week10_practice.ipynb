- Unit_Price
- Total_Sale
- Sales_Rep
- Region
Write beginner-friendly pandas code that shows, for each region:

Sort by total sales (highest first), format dollar values with $ and commas,
and use clear column names suitable for a manager presentation.(make sure the data is cleaned )
Add a comment explaining each step.
```]
```

**AI Response:**

````python
# Paste AI-generated code here and run it
import pandas as pd

# =====================================================
# STEP 1: CLEAN THE DATA
# =====================================================

# Work on a copy so the original data stays untouched
clean_df = sales_df.copy()

# Remove any accidental spaces from the column names
clean_df.columns = clean_df.columns.str.strip()

# The number columns were saved as text with "$" signs (even Quantity!).
# Remove "$" and "," then convert the text to real numbers.
# errors='coerce' turns anything unreadable into NaN (blank) instead of crashing.
for col in ['Quantity', 'Unit_Price', 'Total_Sale']:
    clean_df[col] = (clean_df[col].astype(str)
                     .str.replace('$', '', regex=False)
                     .str.replace(',', '', regex=False)
                     .str.strip())
    clean_df[col] = pd.to_numeric(clean_df[col], errors='coerce')

# Convert the Date column from text to a real date
clean_df['Date'] = pd.to_datetime(clean_df['Date'], errors='coerce')

# Tidy the text columns: remove extra spaces, make region codes uppercase
# (so "nsw " and "NSW" are counted as the same region)
clean_df['Region'] = clean_df['Region'].astype(str).str.strip().str.upper()
clean_df['Sales_Rep'] = clean_df['Sales_Rep'].str.strip()
clean_df['Product'] = clean_df['Product'].str.strip()

# Remove duplicate rows and rows missing a sale amount or region
rows_before = len(clean_df)
clean_df = clean_df.drop_duplicates()
clean_df = clean_df.dropna(subset=['Total_Sale', 'Region'])
rows_after = len(clean_df)
print(f"Cleaning: {rows_before} rows in, {rows_after} rows kept, "
      f"{rows_before - rows_after} removed")

# Sanity check: Quantity x Unit_Price should equal Total_Sale
mismatch = clean_df[(clean_df['Quantity'] * clean_df['Unit_Price']
                     - clean_df['Total_Sale']).abs() > 0.01]
print(f"Rows where Quantity x Unit_Price does not equal Total_Sale: {len(mismatch)}")

# =====================================================
# STEP 2: CALCULATE THE FIGURES FOR EACH REGION
# =====================================================

# Group rows by Region, then calculate the sum, average and count of Total_Sale
region_summary = clean_df.groupby('Region').agg(
    total_sales=('Total_Sale', 'sum'),
    average_sale=('Total_Sale', 'mean'),
    transactions=('Total_Sale', 'count')
)

# Each region's share = its total / the company total x 100
region_summary['share_pct'] = (region_summary['total_sales']
                               / region_summary['total_sales'].sum() * 100)

# Sort so the highest-selling region comes first
region_summary = region_summary.sort_values('total_sales', ascending=False)

# =====================================================
# STEP 3: FORMAT FOR THE MANAGER PRESENTATION
# =====================================================

# Build a display table with clear column names.
# '${:,.2f}' means: add $, add thousands commas, show 2 decimal places.
report = pd.DataFrame({
    'Region': region_summary.index,
    'Total Sales': region_summary['total_sales'].map('${:,.2f}'.format),
    'Average Sale': region_summary['average_sale'].map('${:,.2f}'.format),
    'No. of Transactions': region_summary['transactions'],
    'Share of Total Sales': region_summary['share_pct'].map('{:.1f}%'.format),
}).reset_index(drop=True)

# Number the rows 1, 2, 3... (a ranking) instead of starting at 0
report.index = report.index + 1

# Print the final table
print("\n📊 Sales Performance by Region")
print("=" * 70)
print(report.to_string())
print("=" * 70)
print(f"Grand total: ${region_summary['total_sales'].sum():,.2f} "
      f"from {region_summary['transactions'].sum()} transactions")
# Remember to critique and improve if needed!
````

## Practice Exercise 2.1: Spending Trend Analysis

::: {.callout-tip}
## Business Context

You're helping a friend understand their spending habits over time.

**Challenge**: Ask AI to help you:
1. Convert dates to datetime format
2. Analyze spending trends by day of week
3. Identify if spending has increased over time
4. Spot any unusual spending patterns

**Focus**: Practice asking AI to explain any complex date handling code it provides.
:::

**Your Analysis:**

** Prompt:**
```
I'm helping a friend understand their spending. transactions_df has columns
Date, Amount (refunds are negative), Category, Description.
Write beginner-friendly pandas code that:
1. Converts Date to datetime
2. Shows total spending by day of week (Monday to Sunday order)
3. Checks whether spending went up over time (compare first half vs second half)
4. Flags unusual spending patterns
Exclude refunds from the spending figures. Explain every date-related line
in a comment, since I'm new to datetime.
```

````python
import pandas as pd

# ---------- STEP 1: CLEAN THE DATA ----------
# Work on a copy so the original stays untouched
clean_tx = transactions_df.copy()

# Amount is text like "$45.50" or "-$25.00": remove "$", then convert to numbers
clean_tx['Amount'] = clean_tx['Amount'].astype(str).str.replace('$', '', regex=False)
clean_tx['Amount'] = pd.to_numeric(clean_tx['Amount'], errors='coerce')

# TASK 1: CONVERT DATE TO DATETIME
# Right now Date is just text, e.g. "2024-08-01". Python can't tell that
# "2024-08-02" comes after "2024-08-01" or what weekday it is.
# pd.to_datetime() turns the text into a real date Python understands.
# errors='coerce' means: if a date can't be read (e.g. "abc"), make it NaT
# ("Not a Time", the date version of a blank) instead of crashing.
clean_tx['Date'] = pd.to_datetime(clean_tx['Date'], errors='coerce')

# Remove duplicate rows and rows where Date (NaT) or Amount could not be read
clean_tx = clean_tx.drop_duplicates().dropna(subset=['Date', 'Amount'])

# EXCLUDE REFUNDS: keep only positive amounts (purchases).
# Refunds are negative, so they are kept separately and NOT counted as spending.
spending = clean_tx[clean_tx['Amount'] > 0].copy()
refunds = clean_tx[clean_tx['Amount'] < 0]

print(f"Purchases: ${spending['Amount'].sum():,.2f} | "
      f"Refunds (excluded): -${abs(refunds['Amount'].sum()):,.2f}\n")

# ---------- TASK 2: SPENDING BY DAY OF WEEK ----------
# .dt gives access to date tools on a date column.
# .dt.day_name() reads each date and returns its weekday as a word,
# e.g. 2024-08-01 -> "Thursday". We store it in a new 'Day' column.
spending['Day'] = spending['Date'].dt.day_name()

# groupby sorts days alphabetically (Friday, Monday, Saturday...),
# so we write the correct calendar order ourselves.
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Add up spending per day, .reindex() rearranges rows into day_order,
# and .fillna(0) puts $0 for any day with no purchases.
by_day = spending.groupby('Day')['Amount'].sum().reindex(day_order).fillna(0)

print("📅 Total Spending by Day of Week")
print("=" * 35)
print(by_day.map('${:,.2f}'.format).to_string())
print()

# ---------- TASK 3: FIRST HALF vs SECOND HALF ----------
# .min() finds the EARLIEST date and .max() the LATEST date.
# Subtracting two dates gives a length of time (a "Timedelta"),
# e.g. 20 Aug - 1 Aug = 19 days.
# Dividing that by 2 gives half the period (9.5 days), and adding it to the
# earliest date gives the halfway date.
midpoint = spending['Date'].min() + (spending['Date'].max() - spending['Date'].min()) / 2

# Dates can be compared like numbers: earlier dates are "smaller".
# <= midpoint keeps dates on or before the halfway point (first half);
# >  midpoint keeps dates after it (second half).
first_half = spending[spending['Date'] <= midpoint]['Amount'].sum()
second_half = spending[spending['Date'] > midpoint]['Amount'].sum()

# Percentage change from the first half to the second half
change_pct = (second_half - first_half) / first_half * 100

print("📈 Spending Over Time")
print("=" * 35)
# {midpoint:%d %b} formats the date for display:
# %d = day number (10), %b = short month name (Aug) -> "10 Aug"
print(f"First half  (to {midpoint:%d %b}): ${first_half:,.2f}")
print(f"Second half:             ${second_half:,.2f}")
print(f"Change: {change_pct:+.1f}%  ->  ", end="")
if change_pct > 10:
    print("spending went UP")
elif change_pct < -10:
    print("spending went DOWN")
else:
    print("spending stayed roughly the SAME")
print()

# ---------- TASK 4: FLAG UNUSUAL SPENDING ----------
# A purchase is "unusual" if it is more than 2 standard deviations above average
# (standard deviation = how spread out the amounts typically are)
threshold = spending['Amount'].mean() + 2 * spending['Amount'].std()
unusual = spending[spending['Amount'] > threshold]

print("🚩 Unusually Large Purchases")
print("=" * 35)
print(f"(anything over ${threshold:,.2f})")
print(unusual[['Date', 'Category', 'Description', 'Amount']].to_string(index=False))
print()

# Small purchases that happen often can add up (5+ purchases, under $20 each on average)
by_category = spending.groupby('Category')['Amount'].agg(['count', 'mean', 'sum'])
habits = by_category[(by_category['count'] >= 5) & (by_category['mean'] < 20)]

print("☕ Frequent Small Purchases That Add Up")
print("=" * 35)
print(habits.rename(columns={'count': 'Purchases', 'mean': 'Average', 'sum': 'Total'}).round(2).to_string())
````

## Practice Exercise 2.2: Budget Category Analysis

**Advanced**: Ask AI to explain the statistical concepts it uses.

**Your Budget Analysis:**

````python
import pandas as pd
import math

# ---------- STEP 1: CLEAN THE DATA ----------

# Work on a copy so the original stays untouched
budget_tx = transactions_df.copy()

# Amount is text like "$45.50": remove "$" and make it a number
amount_text = budget_tx['Amount'].astype(str)
amount_text = amount_text.str.replace('$', '', regex=False)
budget_tx['Amount'] = pd.to_numeric(amount_text, errors='coerce')

# Date is text like "2024-08-01": make it a real date
budget_tx['Date'] = pd.to_datetime(budget_tx['Date'], errors='coerce')

# Remove duplicate rows and rows that could not be read
budget_tx = budget_tx.drop_duplicates()
budget_tx = budget_tx.dropna(subset=['Date', 'Amount'])

# Leave out refunds: keep positive amounts only
spending = budget_tx[budget_tx['Amount'] > 0].copy()

# How many days the data covers (+1 so 1-20 Aug = 20 days)
first_day = spending['Date'].min()
last_day = spending['Date'].max()
days_covered = (last_day - first_day).days + 1

# How many days are in that month (August = 31)
days_in_month = last_day.days_in_month

print("Days of data:", days_covered, "out of", days_in_month)
print()

# ---------- STEP 2: SPENDING PER CATEGORY (TASK 1) ----------

# For each category: total, count, average and standard deviation
summary = spending.groupby('Category')['Amount'].agg(
    ['sum', 'count', 'mean', 'std'])
summary = summary.sort_values('sum', ascending=False)

print("TABLE 1: Monthly Spending by Category")
print("=" * 45)
table1 = pd.DataFrame()
table1['Spent This Month'] = summary['sum'].map('${:,.2f}'.format)
table1['Purchases'] = summary['count']
print(table1.to_string())
print()
print("Total spent: $" + format(summary['sum'].sum(), ',.2f'))
print("Note: there is only one month of data, so the")
print("average monthly spend is just this month's total.")
print()

# ---------- STEP 3: HOW UP-AND-DOWN IS EACH CATEGORY? (TASK 2) ----------

# STANDARD DEVIATION in plain English:
# It shows how far purchases usually are from the average.
#  LOW  = purchases are similar in size -> easy to budget
#  HIGH = purchase sizes jump around    -> harder to budget
# We compare it to the average purchase, because $5 of
# variation is big for coffee but tiny for groceries.

def consistency_label(row):
    if row['count'] < 2:
        return 'Only 1 purchase'
    ratio = row['std'] / row['mean']
    if ratio < 0.25:
        return 'Steady'
    elif ratio < 0.6:
        return 'Somewhat up-and-down'
    else:
        return 'Very up-and-down'

summary['consistency'] = summary.apply(consistency_label, axis=1)

print("TABLE 2: How Consistent Is Each Category?")
print("=" * 70)
table2 = pd.DataFrame()
table2['Average Purchase'] = summary['mean'].map('${:,.2f}'.format)
table2['Std Deviation'] = summary['std'].round(2).fillna(0)
table2['Consistency'] = summary['consistency']
print(table2.to_string())
print()

# ---------- STEP 4: SUGGESTED MONTHLY BUDGET (TASK 3) ----------

# How the budget is worked out:
# 1. Fixed bills (electricity) are kept as they are.
# 2. Other spending only covers part of the month, so we
#    stretch it: spent / days covered x days in month.
# 3. Add a 10% buffer for surprises.
# 4. Round UP to the nearest $10.

fixed_bills = ['Utilities']

estimates = []
for category, spent in summary['sum'].items():
    if category in fixed_bills:
        estimates.append(spent)
    else:
        estimates.append(spent / days_covered * days_in_month)

summary['estimate'] = estimates
with_buffer = summary['estimate'] * 1.10
summary['budget'] = (with_buffer / 10).apply(math.ceil) * 10

print("TABLE 3: Suggested Monthly Budget")
print("=" * 60)
table3 = pd.DataFrame()
table3['Spent So Far'] = summary['sum'].map('${:,.2f}'.format)
table3['Full Month'] = summary['estimate'].map('${:,.2f}'.format)
table3['Budget'] = summary['budget'].map('${:,.0f}'.format)
print(table3.to_string())
print()
print("Total monthly budget: $" + format(summary['budget'].sum(), ',.0f'))
print("Utilities kept as a fixed bill; others stretched to 31 days.")
print()

# ---------- STEP 5: WHERE COULD SPENDING BE CUT? (TASK 4) ----------

# "Wants" are easier to cut than "needs" (groceries, transport, bills)
wants = ['Entertainment', 'Dining', 'Coffee']

reasons = {}
reasons['Entertainment'] = 'Optional - check subscriptions'
reasons['Dining'] = 'Cook at home 1-2 more times'
reasons['Coffee'] = 'Small but frequent - adds up'

cut = summary[summary.index.isin(wants)].copy()
cut = cut.sort_values('estimate', ascending=False)

# Saving if each "want" is cut by a quarter (25%)
cut['saving'] = cut['estimate'] * 0.25

print("TABLE 4: Most Realistic Places to Cut Back")
print("=" * 70)
table4 = pd.DataFrame()
table4['Full Month'] = cut['estimate'].map('${:,.2f}'.format)
table4['Save if Cut 25%'] = cut['saving'].map('${:,.2f}'.format)
table4['Why'] = [reasons[c] for c in cut.index]
print(table4.to_string())
print()

total_saving = cut['saving'].sum()
print("Possible saving: $" + format(total_saving, ',.2f') + " a month")
print("Groceries, Transport and Utilities are needs, so not listed.")
````

## Scenario 1: Retail Store Analysis

::: {.callout-note}
## Business Context

You're analyzing sales data for a small retail chain. They want to know:
- Which stores need more inventory
- What products to discontinue
- When to schedule promotional campaigns

**Data**: Use the sales_data.csv but imagine it represents different store locations instead of regions.

**Your Business Analysis:**

**Prompt:**
```
I'm analysing sales for a small retail chain. sales_df has columns Date,
Product, Quantity, Unit_Price, Total_Sale, Sales_Rep and Region - treat each
Region as a store location. The number columns are text with $ signs, so
clean them first.

Write beginner-friendly pandas code that answers four questions, one table each:
1. Which stores need more inventory? (rank stores by units sold and show
   each store's best-selling product)
2. Which products could be discontinued? (bottom 5 products by revenue,
   with units sold and how many stores sell them)
3. When should promotions be scheduled? (revenue and number of sales by day
   of week, Monday to Sunday; point out the quietest days)
4. How should sales staff be allocated? (reps per store, sales per rep and
   revenue per rep)

After each table, print a one-line takeaway. Warn me where the data is too
limited to be sure (for example, no cost data or only a few weeks of dates).
Format money with $ and commas, and add a short comment for each step.
```

````python
# Practice comprehensive business analysis with AI assistance

# Focus on actionable insights, not just data summaries
import pandas as pd

# ---------- STEP 1: CLEAN THE DATA ----------
# Same cleaning as Exercise 1.1, on a fresh copy
store_df = sales_df.copy()
store_df.columns = store_df.columns.str.strip()

# Remove "$" and "," then convert the number columns to real numbers
for col in ['Quantity', 'Unit_Price', 'Total_Sale']:
    store_df[col] = (store_df[col].astype(str)
                     .str.replace('$', '', regex=False)
                     .str.replace(',', '', regex=False)
                     .str.strip())
    store_df[col] = pd.to_numeric(store_df[col], errors='coerce')

store_df['Date'] = pd.to_datetime(store_df['Date'], errors='coerce')

# The scenario says to treat each Region as a store location
store_df['Store'] = store_df['Region'].astype(str).str.strip().str.upper()
store_df['Product'] = store_df['Product'].astype(str).str.strip()
store_df['Sales_Rep'] = store_df['Sales_Rep'].astype(str).str.strip().str.title()

store_df = store_df.drop_duplicates()
store_df = store_df.dropna(subset=['Date', 'Store', 'Product', 'Quantity', 'Total_Sale'])

print(f"Data: {len(store_df)} sales across {store_df['Store'].nunique()} stores, "
      f"{store_df['Date'].min():%d %b %Y} to {store_df['Date'].max():%d %b %Y}")
print()

# ---------- Q1: WHICH STORES NEED MORE INVENTORY? ----------
# Stores that move the most units use up stock fastest
stores = store_df.groupby('Store').agg(
    units_sold=('Quantity', 'sum'),
    revenue=('Total_Sale', 'sum'),
    sales=('Total_Sale', 'count')
).sort_values('units_sold', ascending=False)

# Best-selling product (by units) in each store
units_by_store_product = store_df.groupby(['Store', 'Product'])['Quantity'].sum().reset_index()
best = (units_by_store_product.sort_values('Quantity', ascending=False)
        .drop_duplicates('Store')        # keep only the top product for each store
        .set_index('Store'))
stores['best_seller'] = best['Product']

table1 = pd.DataFrame({
    'Store': stores.index,
    'Units Sold': stores['units_sold'].astype(int),
    'Revenue': stores['revenue'].map('${:,.2f}'.format),
    'No. of Sales': stores['sales'],
    'Best Seller (units)': stores['best_seller'],
}).reset_index(drop=True)
table1.index = table1.index + 1   # rank from 1 instead of 0

print("📦 Q1: Stores Ranked by Units Sold (highest stock usage first)")
print("=" * 80)
print(table1.to_string())
print(f"-> {stores.index[0]} moves the most stock: prioritise restocking there, "
      f"especially {stores['best_seller'].iloc[0]}.")
print()

# ---------- Q2: WHAT PRODUCTS COULD BE DISCONTINUED? ----------
products = store_df.groupby('Product').agg(
    revenue=('Total_Sale', 'sum'),
    units_sold=('Quantity', 'sum'),
    sales=('Total_Sale', 'count'),
    stores=('Store', 'nunique')          # how many stores sell it
)

# Lowest revenue first = weakest products
weakest = products.sort_values(['revenue', 'units_sold']).head(5)

table2 = pd.DataFrame({
    'Product': weakest.index,
    'Revenue': weakest['revenue'].map('${:,.2f}'.format),
    'Units Sold': weakest['units_sold'].astype(int),
    'No. of Sales': weakest['sales'],
    'Stores Selling It': weakest['stores'],
}).reset_index(drop=True)
table2.index = table2.index + 1

print("🗑️  Q2: Bottom 5 Products by Revenue (candidates to review, not auto-cut)")
print("=" * 80)
print(table2.to_string())
print("-> Before discontinuing, check profit margin and whether these products")
print("   are add-ons that bring in bigger sales (no cost data here to confirm).")
print()

# ---------- Q3: WHEN TO SCHEDULE PROMOTIONS? ----------
# .dt.day_name() turns each date into its weekday name, e.g. "Monday"
store_df['Day'] = store_df['Date'].dt.day_name()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

by_day = store_df.groupby('Day').agg(
    revenue=('Total_Sale', 'sum'),
    sales=('Total_Sale', 'count')
).reindex(day_order).fillna(0)          # calendar order, $0 for days with no sales

table3 = pd.DataFrame({
    'Day': by_day.index,
    'Revenue': by_day['revenue'].map('${:,.2f}'.format),
    'No. of Sales': by_day['sales'].astype(int),
}).reset_index(drop=True)

print("📅 Q3: Sales by Day of Week")
print("=" * 80)
print(table3.to_string(index=False))

# Promote on quiet days to lift them; don't discount days that already sell well
quiet_days = by_day[by_day['sales'] > 0].sort_values('revenue').head(2).index.tolist()
busiest_day = by_day['revenue'].idxmax()
print(f"-> Quietest trading days: {', '.join(quiet_days)} - best for promotions.")
print(f"-> Busiest day: {busiest_day} - customers already come, so no discount needed.")

# How many days the data covers (+1 so the first and last day both count)
days_span = (store_df['Date'].max() - store_df['Date'].min()).days + 1
if days_span < 60:
    print(f"   Caution: only {days_span} days of data, so seasonal timing can't be judged.")
print()

# ---------- Q4: HOW TO ALLOCATE SALES STAFF? ----------
staff = store_df.groupby('Store').agg(
    reps=('Sales_Rep', 'nunique'),       # number of different reps
    sales=('Total_Sale', 'count'),
    revenue=('Total_Sale', 'sum')
)
# Workload = sales handled per rep; a high number means staff are stretched
staff['sales_per_rep'] = staff['sales'] / staff['reps']
staff['revenue_per_rep'] = staff['revenue'] / staff['reps']
staff = staff.sort_values('sales_per_rep', ascending=False)

table4 = pd.DataFrame({
    'Store': staff.index,
    'Reps': staff['reps'],
    'No. of Sales': staff['sales'],
    'Sales per Rep': staff['sales_per_rep'].round(1),
    'Revenue per Rep': staff['revenue_per_rep'].map('${:,.2f}'.format),
}).reset_index(drop=True)
table4.index = table4.index + 1

print("👥 Q4: Staff Workload by Store (busiest per rep first)")
print("=" * 80)
print(table4.to_string())
print(f"-> {staff.index[0]} has the most sales per rep: first candidate for extra staff.")
print(f"-> {staff.index[-1]} has the lightest workload per rep: could share staff if needed.")
print()
print(f"Note: based on {len(store_df)} sales only. Treat these as starting points to")
print("discuss with store managers, not final decisions.")


````

## Scenario 2: Personal Finance Coaching

::: {.callout-note}
## Business Context

You're building a personal finance app feature that provides spending advice.

**Challenge**: Use AI to create code that:
1. Identifies spending categories that are above typical percentages
2. Suggests specific areas for potential savings
3. Provides encouraging feedback for good spending habits
4. Formats advice in a user-friendly way
:::

**Your Coaching Algorithm:**

````python
# AI-assisted personal finance advice generation

import pandas as pd

# ---------- STEP 1: CLEAN THE DATA ----------
coach_tx = transactions_df.copy()
# Amount may already be a number (from the reload cell) or text with "$" - this handles both
coach_tx['Amount'] = pd.to_numeric(
    coach_tx['Amount'].astype(str).str.replace('$', '', regex=False), errors='coerce')
coach_tx['Date'] = pd.to_datetime(coach_tx['Date'], errors='coerce')
coach_tx = coach_tx.drop_duplicates().dropna(subset=['Date', 'Amount', 'Category'])

# Purchases only - refunds (negative amounts) are not spending
spend = coach_tx[coach_tx['Amount'] > 0]

# ---------- STEP 2: COMPARE WITH TYPICAL PERCENTAGES ----------
# "Typical" share of total spending for each category.
# These are EXAMPLE assumptions for this exercise, not official figures.
# A real app would use benchmarks from a trusted source (e.g. ABS household data).
typical_share = {
    'Groceries': 35, 'Transport': 15, 'Utilities': 15,
    'Dining': 10, 'Entertainment': 10, 'Coffee': 3,
}
leeway = 2   # allow 2 percentage points above typical before flagging

by_cat = spend.groupby('Category')['Amount'].agg(['sum', 'count'])
total = by_cat['sum'].sum()
by_cat['share'] = by_cat['sum'] / total * 100            # % of all spending
by_cat['typical'] = by_cat.index.map(typical_share)      # blank (NaN) if no benchmark
by_cat['gap'] = by_cat['share'] - by_cat['typical']      # positive = above typical
by_cat = by_cat.sort_values('gap', ascending=False)

# ---------- STEP 3: ONE SAVINGS TIP PER CATEGORY ----------
tips = {
    'Groceries': 'Plan meals and shop with a list; compare unit prices.',
    'Transport': 'Check whether a weekly transit pass or car-pooling is cheaper.',
    'Utilities': 'Compare energy plans; small habit changes add up.',
    'Dining': 'Swap one or two meals out per week for home cooking.',
    'Entertainment': 'Review subscriptions and cancel ones you rarely use.',
    'Coffee': 'Make coffee at home on some days - small buys add up.',
}

# Split categories into groups (rows with no benchmark fall out of both comparisons)
overspend = by_cat[by_cat['gap'] > leeway]
on_track = by_cat[by_cat['gap'] <= leeway]
no_benchmark = by_cat[by_cat['typical'].isna()]

# ---------- STEP 4: PRINT FRIENDLY ADVICE ----------
print("💬 Your Spending Check-In")
print("=" * 60)
print(f"Total spending: ${total:,.2f} across {len(by_cat)} categories")
print()

print("⚠️  Areas to look at")
print("-" * 60)
total_saving = 0
if overspend.empty:
    print("Nothing stands out - all categories are close to typical.")
for category, row in overspend.iterrows():
    # Approximate saving if this category matched the typical share
    saving = (row['share'] - row['typical']) / 100 * total
    total_saving += saving
    print(f"{category}: {row['share']:.1f}% of spending (typical ~{row['typical']:.0f}%)")
    print(f"   Tip: {tips.get(category, 'Look for cheaper options here.')}")
    print(f"   Bringing it to typical would free up about ${saving:,.2f}")
print()

print("✅ Doing well")
print("-" * 60)
if on_track.empty:
    print("Every category is above typical - start with the biggest gap above.")
for category, row in on_track.iterrows():
    print(f"{category}: {row['share']:.1f}% (typical ~{row['typical']:.0f}%) - nicely under control!")
print()

if not no_benchmark.empty:
    print("ℹ️  No benchmark yet for: " + ', '.join(no_benchmark.index))
    print()

print("=" * 60)
if total_saving > 0:
    print(f"Possible saving: about ${total_saving:,.2f} for this period.")
print("Small, steady changes beat big cuts you can't keep. You've got this! 🎉")
print("Note: shares are % of total spending, not income - income isn't in this data.")
````
